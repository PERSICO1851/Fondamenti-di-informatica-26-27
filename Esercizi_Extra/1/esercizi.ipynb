```python
# Esercizi per casa (Assegnamenti, Condizioni e Cicli) - Esercizio 5: Calcolatrice con match
#
# Sviluppa un programma che:
# - Chiede due numeri interi a e b e un operatore tra +, -, *, x, /, //, %, **
# - Usa l'istruzione match sull'operatore e stampa l'operazione completa con il suo risultato
# - Accetta sia * che x per il prodotto (usando | in un unico case)
# - Per /, // e % controlla prima che b non sia 0: in quel caso stampa un messaggio di errore
# - Per qualsiasi altro operatore stampa che non è valido (caso di default _)
#
# Esempi di output (esecuzioni diverse):
# Primo numero: 7
# Secondo numero: 2
# Operatore (+ - * x / // % **): //
# 7 // 2 = 3
#
# Primo numero: 3
# Secondo numero: 4
# Operatore (+ - * x / // % **): x
# 3 * 4 = 12
#
# Primo numero: 10
# Secondo numero: 0
# Operatore (+ - * x / // % **): /
# Errore: divisione per zero
#
# Primo numero: 2
# Secondo numero: 5
# Operatore (+ - * x / // % **): ^
# Operatore non valido: ^
a = int(input("numero?"))
b = int(input("numero?"))
operatori = "+ - * / // % **"
operatore = str(input("operatore?"))
if operatore in operatori.strip(" "):
    match operatore:
        case "+":
            print(a+b)
        case "-":
            print(a-b)
        case "*":
            print(a*b)
        case "/":
            if b != 0:
                print(a / b)
            else:
                print("errore Zero Division")
        case "//":
            if b != 0:
                print(a // b)
            else:
                print("errore Zero Division")
        case "%":
            if b != 0:
                print(a % b)
            else:
                print("errore Zero Division")
        case "**":
            print(a+b)
else:
    print(f"Operatore non valido: {operatore}")
```

```python
# Python: Funzioni, Dizionari, Liste, Tuple, Insiemi e Sort - 6. Funzioni - Esercizio 6.1: Funzione con parametri opzionali
#
# Scrivi una funzione che calcoli l'area di un rettangolo. Se viene passato solo
# un parametro, considera il rettangolo come un quadrato.

def area_rettangolo(larghezza, altezza=None):
    # Il tuo codice qui
    if altezza != None:
        return larghezza*altezza
    else:
        return larghezza**2

# Test
print(area_rettangolo(5))     # 25 (quadrato)
print(area_rettangolo(4, 6))  # 24 (rettangolo)
```
