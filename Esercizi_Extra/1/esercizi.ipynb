```python
# Esercizi sui contenitori - Esercizio 1: Lista di Frutta
#
# - Crea la lista frutta = ["mela", "banana", "arancia"].
# - Aggiungi due frutti nuovi con un ciclo for.
# - Stampa tutti gli elementi con il loro indice (1, 2, 3...).
frutta = ["mela", "banana", "arancia"]
for i in range(2):
    frutta.append(f"mela{i}")
for y in frutta:
    print(frutta.index(y)+1)
```

```python
# Esercizi sui contenitori - Esercizio 2: Voti di Studenti
#
# - Partendo da un dizionario vuoto voti = {}, chiedi nome e voto (1-10) finché
#   l'utente inserisce "fine".
# - Controlla che il valore sia un numero tra 1 e 10 (altrimenti ripeti la richiesta).
# - Salva ogni coppia nome:voto nel dizionario.
# - Alla fine, stampa quanti studenti hanno voto >= 6 e la tabella dei nomi e voti.
diz = {}
while True:
    nome = (input("nome?"))
    if nome == "fine":
        break
    voto = int(input("voto?"))
    while not 10 >= voto >= 1:
        print("voto invalido!")
        voto = int(input("voto?"))
    diz[nome] = voto
count = 0
for nome,voto in diz.items():
    print(nome,voto)
    if voto  >= 6:
        count += 1
print(f"Number Students: {count}")
```

```python
# Esercizi sui contenitori - Esercizio 3: Numeri Unici
#
# - Aggiungi dei numeri inseriti da tastiera alla lista nums = [].
# - Continua finché l'utente digita "stop".
# - Stampa i numeri senza duplicati (usa un set) in ordine crescente (usa sort).
numbers = []
while True:
    nums = input("nums?")
    if nums == "stop":
        break
    numbers.append(nums)
numbers.sort()
numbers = set(numbers)
print(numbers)
```
