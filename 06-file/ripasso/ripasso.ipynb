## 1. Dizionari

Un dizionario associa delle **chiavi** a dei **valori**. Si scrive tra parentesi graffe, con coppie `chiave: valore`.

```python
telefoni = {'Luigi': '555-871612', 'Mario': '555-826781'}
print(telefoni)
print(len(telefoni))     # numero di coppie chiave-valore
```

### Leggere un valore

Con `d[chiave]` si ottiene il valore associato alla chiave. Se la chiave non c'è, Python dà un errore (`KeyError`).
Con `d.get(chiave, default)` invece si ottiene `default` quando la chiave manca.
Con `in` si controlla se una chiave è presente.

```python
print(telefoni['Mario'])
print(telefoni.get('Peach', 'non trovato'))
print('Luigi' in telefoni)
print('Peach' in telefoni)

# Togliete il commento alla riga seguente per vedere l'errore
# print(telefoni['Peach'])
```

### Aggiungere, modificare e togliere

Assegnare a una chiave nuova aggiunge una coppia; assegnare a una chiave esistente ne cambia il valore. `del` toglie una coppia.

```python
telefoni['Peach'] = '555-000111'     # aggiunge
telefoni['Mario'] = '555-1234'       # modifica
del telefoni['Luigi']                # toglie
print(telefoni)
```

### Scorrere un dizionario

- `for k in d` scorre le chiavi
- `d.values()` dà i valori
- `d.items()` dà le coppie `(chiave, valore)`

```python
spese = {'Luigi': 100, 'Mario': 1000, 'Peach': 250}

for nome in spese:
    print(nome)

print(sum(spese.values()))

for nome, importo in spese.items():
    print(nome, 'ha speso', importo)
```

### Contare con un dizionario

Uno schema molto comune: usare un dizionario per contare quante volte compare ogni elemento.
`conteggi.get(x, 0)` vale 0 la prima volta che incontriamo `x`.

```python
frutti = ['mela', 'pera', 'mela', 'kiwi', 'mela', 'pera']

conteggi = {}
for frutto in frutti:
    conteggi[frutto] = conteggi.get(frutto, 0) + 1
print(conteggi)
```

## 2. `split` e `join`

`s.split()` divide una stringa in una **lista** di parti:

- senza argomenti divide sugli spazi (anche più spazi di fila, tabulazioni e a capo) e scarta le parti vuote;
- con un separatore, ad esempio `s.split(',')`, divide esattamente su quel carattere e **tiene** le parti vuote.

```python
riga = '  Luigi     555-871612   '
print(riga.split())

riga_csv = 'Luigi,,100,10'
print(riga_csv.split(','))      # notate la stringa vuota tra le due virgole
```

### Pulire le stringhe

- `s.strip()` toglie spazi e a capo all'inizio e alla fine
- `s.lower()` e `s.upper()` cambiano maiuscole e minuscole
- `int(s)` converte una stringa di cifre in un intero; `s.isdigit()` dice se la stringa contiene solo cifre

```python
print(repr('  Ciao Mario!\n'.strip()))
print('Ciao Mario!'.lower())
print(int('1000') + 1)
print('1000'.isdigit(), ''.isdigit(), '555-1234'.isdigit())
```

### `join`

`separatore.join(lista)` fa l'operazione inversa di `split`: unisce una lista di stringhe mettendo `separatore` in mezzo.
Gli elementi devono essere stringhe.

```python
nomi = ['Luigi', 'Mario', 'Peach']
print(', '.join(nomi))
print('-'.join(['2026', '11', '03']))

numeri = [1, 2, 3]
print('+'.join(str(n) for n in numeri))   # prima convertiamo i numeri in stringhe
```

## 3. File

### Leggere un file

Si apre un file con `open(nome_file)`. Conviene usarlo insieme a `with`: il file viene chiuso automaticamente alla fine del blocco.
`f.read()` restituisce tutto il contenuto come una sola stringa.

Il nome del file è un percorso **relativo** alla cartella da cui eseguite il codice: per questo il notebook va aperto dalla cartella `06-file/ripasso`.

```python
with open('file01.txt', encoding='utf-8') as f:
    testo = f.read()
print(testo)
print(type(testo), len(testo))
```

### Leggere riga per riga

- scorrendo il file con `for riga in f`, ogni riga **include** l'a capo finale (`'\n'`): spesso serve `riga.strip()`
- `testo.splitlines()` divide il testo in righe, **senza** il carattere di a capo


```python
with open('file02.txt', encoding='utf-8') as f:
    for riga in f:
        #print(repr(riga))
        print(riga.strip())

print()

with open('file02.txt', encoding='utf-8') as f:
    righe = f.read().splitlines()
print(righe)
```

### Scrivere un file

Con `open(nome_file, 'w')` si apre un file in scrittura: se esiste già, il contenuto **viene cancellato**.
Con `'a'` (append) si aggiunge in fondo. `f.write(s)` non aggiunge l'a capo: dobbiamo metterlo noi con `'\n'`.

```python
with open('ripasso_prova.txt', 'w', encoding='utf-8') as f:
    f.write('prima riga\n')
    f.write('seconda riga\n')

with open('ripasso_prova.txt', 'a', encoding='utf-8') as f:
    f.write('riga aggiunta\n')

with open('ripasso_prova.txt', encoding='utf-8') as f:
    print(f.read())
```
